```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load Data
url = 'http://becomingvisual.com/python4data/tv.csv'
df = pd.read_csv(url)

# 2. Statistika Deskriptif menggunakan pandas
print("=== STATISTIKA DESKRIPTIF ===")
display(df.describe())

# 3. Scatter Plot dan Boxplot menggunakan pandas
# Scatter Plot
ax_scatter = df.plot.scatter(x='av_rating', y='seasonNumber', alpha=0.5, figsize=(10, 5), color='teal')
ax_scatter.set_title('Hubungan Rating vs Jumlah Musim (Scatter Plot)')
plt.show()

# Boxplot (Mengelompokkan rating berdasarkan jumlah musim untuk melihat hubungannya)
ax_box = df.boxplot(column='av_rating', by='seasonNumber', figsize=(12, 6))
plt.title('Distribusi Rating per Jumlah Musim (Boxplot)')
plt.suptitle('') # Menghapus default title pandas
plt.xlabel('Jumlah Musim')
plt.ylabel('Rata-rata Rating')
plt.xticks(rotation=90)
plt.show()
```

### 4. Insight yang Ditemukan:

*   **Konsentrasi Data**: Mayoritas acara TV memiliki jumlah musim yang sedikit (1-5 musim). Sangat sedikit acara yang bisa mencapai lebih dari 10 musim.
*   **Kualitas vs Umur**: Dari scatter plot terlihat bahwa acara dengan jumlah musim yang sangat banyak (misal > 20) cenderung memiliki rating yang stabil dan tinggi (di atas 7.5). Ini menunjukkan hanya acara berkualitas tinggi yang biasanya dipertahankan hingga puluhan musim.
*   **Rentang Rating**: Rating rata-rata sebagian besar berkumpul di angka 7.5 hingga 8.5.
*   **Outlier**: Terdapat beberapa 'outlier' atau pencilan pada jumlah musim (seperti acara dengan 40+ musim) yang menunjukkan adanya program TV yang sangat ikonik dan berumur panjang meskipun jumlahnya sangat langka.

```python
"""
=======================================================
  TUGAS KE-1 PENAMBANGAN DATA
  Dataset : TV Shows — tv.csv
=======================================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# ── STYLE ──────────────────────────────────────────
sns.set_theme(style='whitegrid', font_scale=1.1)
PALETTE = ['#E63946', '#457B9D', '#F4A261', '#2A9D8F', '#264653']
plt.rcParams.update({'figure.facecolor': '#F8F9FA', 'axes.facecolor': '#FFFFFF'})

# ── LOAD DATA ──────────────────────────────────────
URL = 'http://becomingvisual.com/python4data/tv.csv'
df  = pd.read_csv(URL)

print("=" * 55)
print("  TUGAS KE-1 PENAMBANGAN DATA — TV Shows Dataset")
print("=" * 55)
print(f"\nShape  : {df.shape[0]} baris x {df.shape[1]} kolom")
print(f"Kolom  : {list(df.columns)}")
print(f"Missing:\n{df.isnull().sum().to_string()}")

# ── STATISTIKA DESKRIPTIF ──────────────────────────
print("\n" + "-" * 55)
print("  STATISTIKA DESKRIPTIF")
print("-" * 55)
print(df.describe(include='all').round(3).to_string())

# ── DETEKSI KOLOM ──────────────────────────────────
def find_col(df, keywords):
    for kw in keywords:
        for c in df.columns:
            if kw.lower() in c.lower():
                return c
    return df.select_dtypes(include=np.number).columns[0]

col_r = find_col(df, ['rating', 'avg', 'score'])
col_s = find_col(df, ['season'])

df_plot = df[[col_r, col_s]].dropna()
df_plot.columns = ['rating', 'seasons']
df_plot = df_plot[(df_plot['rating'] > 0) & (df_plot['seasons'] > 0)]

# Regresi
slope, intercept, r, p, _ = stats.linregress(df_plot['rating'], df_plot['seasons'])

# ── FIGURE : 2x3 GRID ─────────────────────────────
fig = plt.figure(figsize=(18, 10), facecolor='#F8F9FA')
fig.suptitle('TV Shows — Analisis Penambangan Data',
             fontsize=20, fontweight='bold', color='#1D3557', y=1.01)
gs = gridspec.GridSpec(2, 3, figure=fig, hspace=0.45, wspace=0.38)

# ── Panel 1: Scatter + Regresi ─────────────────────
ax1 = fig.add_subplot(gs[0, 0])
ax1.scatter(df_plot['rating'], df_plot['seasons'],
            color=PALETTE[0], alpha=0.5, s=50, edgecolors='white', lw=0.5)
x_line = np.linspace(df_plot['rating'].min(), df_plot['rating'].max(), 200)
ax1.plot(x_line, slope * x_line + intercept,
         color=PALETTE[1], lw=2.2, ls='--',
         label=f'R2={r**2:.3f}  p={p:.3f}')
ax1.set_title('Scatter: Rating vs Seasons', fontweight='bold')
ax1.set_xlabel(col_r); ax1.set_ylabel(col_s)
ax1.legend(fontsize=9)

# ── Panel 2: Scatter + KDE 2D ─────────────────────
ax2 = fig.add_subplot(gs[0, 1])
ax2.scatter(df_plot['rating'], df_plot['seasons'],
            color=PALETTE[0], alpha=0.35, s=40, edgecolors='white', lw=0.4)
sns.kdeplot(data=df_plot, x='rating', y='seasons',
            levels=5, fill=True, alpha=0.3, cmap='Reds', ax=ax2)
sns.kdeplot(data=df_plot, x='rating', y='seasons',
            levels=5, color=PALETTE[3], lw=1.2, ax=ax2)
ax2.set_title('Scatter + Density KDE', fontweight='bold')
ax2.set_xlabel(col_r); ax2.set_ylabel(col_s)

# ── Panel 3: Distribusi Rating ─────────────────────
ax3 = fig.add_subplot(gs[0, 2])
ax3.hist(df_plot['rating'], bins=25, color=PALETTE[0], alpha=0.55,
         edgecolor='white', density=True)
df_plot['rating'].plot.kde(ax=ax3, color=PALETTE[1], lw=2.2)
ax3.axvline(df_plot['rating'].mean(),   color=PALETTE[2], lw=2, ls='--',
            label=f"Mean={df_plot['rating'].mean():.2f}")
ax3.axvline(df_plot['rating'].median(), color=PALETTE[3], lw=2, ls=':',
            label=f"Median={df_plot['rating'].median():.2f}")
ax3.set_title('Distribusi Rating', fontweight='bold')
ax3.set_xlabel(col_r); ax3.set_ylabel('Density')
ax3.legend(fontsize=9)

# ── Panel 4: Boxplot Rating per bucket Seasons ─────
ax4 = fig.add_subplot(gs[1, 0])
bins   = [0, 1, 3, 5, 10, 999]
labels = ['1', '2-3', '4-5', '6-10', '10+']
df_plot['cat'] = pd.cut(df_plot['seasons'], bins=bins, labels=labels)
sns.boxplot(data=df_plot, x='cat', y='rating', order=labels,
            palette=PALETTE, width=0.55, linewidth=1.4,
            flierprops=dict(marker='o', markersize=3, alpha=0.5), ax=ax4)
sns.stripplot(data=df_plot, x='cat', y='rating', order=labels,
              color='#1D3557', size=2.5, alpha=0.25, jitter=True, ax=ax4)
ax4.set_title('Boxplot Rating per Jumlah Seasons', fontweight='bold')
ax4.set_xlabel('Jumlah Seasons'); ax4.set_ylabel('Rating')

# ── Panel 5: Violin + Box Overlay ──────────────────
ax5 = fig.add_subplot(gs[1, 1])
sns.violinplot(data=df_plot, x='cat', y='seasons', order=labels,
               palette=PALETTE, inner=None, alpha=0.45, linewidth=1.2, ax=ax5)
sns.boxplot(data=df_plot, x='cat', y='seasons', order=labels,
            width=0.15, color='white', linewidth=1.8,
            flierprops=dict(marker='.', markersize=4), ax=ax5)
ax5.set_title('Violin + Box: Distribusi Seasons', fontweight='bold')
ax5.set_xlabel('Jumlah Seasons'); ax5.set_ylabel('Seasons')

# ── Panel 6: Heatmap Korelasi ──────────────────────
ax6 = fig.add_subplot(gs[1, 2])
num_cols = df.select_dtypes(include=np.number).columns
corr = df[num_cols].corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f',
            cmap='RdYlBu_r', center=0, linewidths=1,
            linecolor='white', ax=ax6, annot_kws={'size': 10})
ax6.set_title('Heatmap Korelasi', fontweight='bold')

plt.savefig('tugas_penambangan_data.png', dpi=150, bbox_inches='tight',
            facecolor='#F8F9FA')
plt.show()

print(f"""
{'='*55}
  INSIGHT YANG DITEMUKAN
{'='*55}

1. KONSENTRASI DATA — MAYORITAS ACARA BERUMUR PENDEK
   Sebagian besar acara TV hanya memiliki 1–5 musim.
   Median jumlah musim = {s_med}, artinya 50% acara
   tidak melampaui angka tersebut. Sangat sedikit yang
   sanggup menembus 10+ musim ({long_run} acara).

2. KUALITAS VS UMUR — RATING TINGGI = LEBIH AWET
   Scatter plot menunjukkan acara dengan jumlah musim
   sangat banyak (>20) cenderung memiliki rating stabil
   dan tinggi (≥7.5). Dari data, {high_rated_long} acara
   memenuhi kriteria ini. Hanya acara berkualitas tinggi
   yang biasanya dipertahankan hingga puluhan musim.

3. RENTANG RATING — TERPUSAT DI 7.5–8.5
   Mean rating = {r_mean:.2f} | Median = {r_med:.2f}.
   Distribusi rating cenderung {'miring kanan' if df_plot['rating'].skew() > 0 else 'miring kiri'}
   (skewness={df_plot['rating'].skew():.3f}), artinya mayoritas
   acara memiliki rating yang cukup baik dan hanya sedikit
   yang berada di rating sangat rendah (<6.0).

4. OUTLIER — PROGRAM IKONIK BERUMUR PANJANG
   Terdapat {outlier_40} acara dengan 40+ musim — jumlah yang
   sangat langka namun menarik. Outlier ini umumnya adalah
   soap opera atau program variety show yang telah menjadi
   ikon budaya. Keberadaannya menarik distribusi seasons
   ke kanan dan mempengaruhi nilai mean secara signifikan.

5. KORELASI — LEMAH NAMUN BERMAKNA
   Pearson r={r:.3f} (R²={r**2:.3f}), p={p:.4f}.
   {'Korelasi signifikan secara statistik (p<0.05).' if p < 0.05 else 'Korelasi tidak signifikan (p>0.05).'}
   Meski korelasi tidak kuat, tren positif tetap terlihat:
   acara dengan rating lebih tinggi punya kecenderungan
   untuk diperpanjang lebih banyak musim oleh produser.
{'='*55}
""")
```
